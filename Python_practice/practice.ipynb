What is a for loop?

A for loop is used when you want to repeat something multiple times.

Instead of writing the same line again and again, we use a for loop to do it automatically.

```python
for i in range(1, 6):
    print(i)
```

```python
fruits = ['apple', 'banana', 'cherry']

for fruit in fruits:
    print(fruit)
```

Q1) Print Even Numbers from 1 to N

Problem:
Given an integer N, print all even numbers from 1 to N (space-separated).

Example:
- Input: 10
- Output: 2 4 6 8 10

Write Python code for this (use a for loop).

```python
N = int(input("Enter a number: "))

for i in range(1, N+1):
    if i % 2 == 0:
        print(i)
```

`for i in range(1, N+1):`

This starts a for loop.

range(1, N+1) generates numbers starting from 1 up to N.

N+1 is used because range does not include the ending number.

Each time the loop runs, the next number is stored in i.

Example if N = 6, i will become: 1, 2, 3, 4, 5, 6

# Q2: Find the Sum of First N Natural Numbers
🧠 Problem

Given a number N, find the sum of numbers from 1 to N using a for loop.

Example:
- Input: 5
- Output: 15

Because:
1+2+3+4+5=15

## Algorithm: Sum of First N Natural Numbers

**Steps:**
1. Start
2. Take input `N`
3. Set `total = 0`
4. Loop from 1 to `N`
    - `total = total + current number`
5. Print `total`
6. End


```python
N = int(input("Enter a number: "))
total = 0

for i in range(1, N + 1):
    total += i
print("The sum of the first", N, "numbers is:", total)
```

# Q3: Count Vowels in a String
🧠 Problem

Given a string, count how many vowels are present.

Vowels are: a, e, i, o, u

Example:

- Input: "interview"
- Output: 4

(Because: i, e, i, e)

## Algorithm: Count Vowels in a String

**Steps:**
1. Start
2. Take string input
3. Set `count = 0`
4. For each character in string:
   - If character is a vowel  
     → `count = count + 1`
5. Print `count`
6. End


```python
S = str(input("Enter a string: "))
count = 0

for char in S:
    if char in 'aeiouAEIOU':
        count += 1
print("Number of vowels in the string:", count)
```
